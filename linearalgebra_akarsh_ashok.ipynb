## Task 2: Linear Algebra

In this task, I will create a matrix from my name, construct a transition matrix, calculate matrix powers, and determine the stationary distribution.

```python
import numpy as np
import sympy as sp

print("Libraries loaded successfully!")
```

```python
# Convert my full name into alphabet positions

name = "AKARSHASHOK"

alphabet_values = [ord(letter) - ord('A') + 1 for letter in name]

print("Name:", name)
print("Alphabet values:", alphabet_values)
print("Number of values:", len(alphabet_values))
```

## 2. Constructing the Square Matrix

My full name contains 11 letters, so m = 11.

The largest integer satisfying n² ≤ 11 is n = 3.

Therefore, I use the first 9 numerical values to construct a 3 × 3 matrix row by row. The remaining two values are discarded.

```python
# Find the matrix size
m = len(alphabet_values)
n = int(np.sqrt(m))

# Select the first n² values
selected_values = alphabet_values[:n**2]

# Store the remaining values
discarded_values = alphabet_values[n**2:]

# Create the square matrix
A = np.array(selected_values).reshape(n, n)

print("Number of letters:", m)
print("Matrix size:", n, "x", n)
print("Selected values:", selected_values)
print("Discarded values:", discarded_values)
print("\nOriginal matrix A:")
print(A)
```

## 3. Constructing the Transition Matrix

To create the transition matrix, I divide each element of a column by the sum of that column.

This normalization ensures that every column sums to 1 and all matrix entries are non-negative.

```python
# Calculate the sum of each column
column_sums = A.sum(axis=0)

# Normalize each column to create the transition matrix
P = A / column_sums

print("Column sums:", column_sums)
print("\nTransition matrix P:")
print(P)

print("\nCheck column sums of P:")
print(P.sum(axis=0))
```

## 4. Calculating Matrix Powers

I calculate the second, fifth, and twentieth powers of the transition matrix using NumPy.

These matrix powers help me observe how the transition probabilities change after repeated steps.

```python
# Calculate powers of the transition matrix

P2 = np.linalg.matrix_power(P, 2)
P5 = np.linalg.matrix_power(P, 5)
P20 = np.linalg.matrix_power(P, 20)

print("P squared (P^2):")
print(P2)

print("\nP to the power of 5 (P^5):")
print(P5)

print("\nP to the power of 20 (P^20):")
print(P20)
```

### Observation of Matrix Powers

When I calculate higher powers of the transition matrix, I observe that the columns gradually become similar.

In P², the columns have different values. In P⁵, they become more similar. In P²⁰, all columns are almost identical.

This indicates that the transition matrix converges towards a stable long-term distribution. This distribution is called the stationary distribution.

## 5. Finding the Stationary Distribution

The stationary distribution is an eigenvector of the transition matrix corresponding to the eigenvalue 1.

I use NumPy to calculate the eigenvalues and eigenvectors and identify the stationary distribution.

```python
# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(P)

# Find the eigenvector corresponding to eigenvalue 1
index = np.argmin(np.abs(eigenvalues - 1))

stationary_vector = np.real(eigenvectors[:, index])

# Normalize the vector so that its elements sum to 1
stationary_vector = stationary_vector / np.sum(stationary_vector)

print("Eigenvalues:")
print(eigenvalues)

print("\nStationary distribution:")
print(stationary_vector)

# Verify the stationary distribution
print("\nP multiplied by stationary distribution:")
print(P @ stationary_vector)

print("\nCheck:")
print(np.allclose(P @ stationary_vector, stationary_vector))
```

### Interpretation of the Stationary Distribution

The stationary distribution is approximately:

- State 1: 13.87%
- State 2: 48.96%
- State 3: 37.17%

State 2 has the highest long-term probability, followed by State 3 and State 1.

The stationary distribution represents the stable long-term proportions of the three states. The verification confirms that multiplying the transition matrix by this vector produces the same vector.

Therefore, the stationary distribution is an eigenvector corresponding to the eigenvalue 1.
